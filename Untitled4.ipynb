```python
import numpy as np

p_pass = 0.9
p_not_pass = 1 - p_pass

p_quick_given_pass = 0.6
p_quick_given_not_pass = 0.3

p_quick = (p_quick_given_pass * p_pass 
           + p_quick_given_not_pass * p_not_pass)

# 贝叶斯公式：P(Pass | Quick)
p_pass_given_quick = (p_quick_given_pass * p_pass) / p_quick

print(p_pass_given_quick)
```

```python
!pip install python-ternary

import numpy as np
import matplotlib.pyplot as plt
import ternary  # pip install python-ternary

alpha_prior = np.array([1.0, 1.0, 1.0])
theta = np.random.dirichlet(alpha_prior)
print("Sampled theta from prior:", theta)

n = 10
x = np.random.multinomial(n, theta)
print("Observed counts x:", x)

alpha_posterior = alpha_prior + x
print("Posterior parameters:", alpha_posterior)

num_samples = 2000
samples_prior = np.random.dirichlet(alpha_prior, size=num_samples)
samples_posterior = np.random.dirichlet(alpha_posterior, size=num_samples)

fig, ax = plt.subplots(figsize=(8, 6))  # matplotlib figure
fig, tax = ternary.figure(scale=1)     # ternary axes

tax.scatter(
    samples_prior,
    marker='o',
    color='blue',
    alpha=0.2,
    label='Prior Samples'
)

tax.scatter(
    samples_posterior,
    marker='x',
    color='red',
    alpha=0.2,
    label='Posterior Samples'
)

tax.legend()
tax.set_title("Dirichlet(1,1,1) -> Posterior after observing Multinomial sample")

tax.boundary()  
tax.gridlines(color="black", multiple=0.1)
tax.ticks(axis='lbr', linewidth=1, multiple=0.1)

plt.show()
```
